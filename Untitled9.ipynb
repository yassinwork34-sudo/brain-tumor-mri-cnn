```python
import os, zipfile, random, hashlib
from pathlib import Path
from collections import Counter

import cv2
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from tqdm.auto import tqdm

import tensorflow as tf
from tensorflow import keras
from tensorflow.keras import layers

from sklearn.model_selection import train_test_split
from sklearn.utils.class_weight import compute_class_weight
from sklearn.metrics import classification_report, confusion_matrix, ConfusionMatrixDisplay

SEED = 42
random.seed(SEED)
np.random.seed(SEED)
tf.random.set_seed(SEED)

print("TensorFlow:", tf.__version__)
print("GPU available:", bool(tf.config.list_physical_devices("GPU")))
```

```python
import os, shutil, zipfile, glob
from pathlib import Path
import pandas as pd

# ---- where is the zip? ----
SOURCE = "drive"      # "drive" = read from your Google Drive, "link" = download with the share link
ZIP_PATH = None       # optional exact path, e.g. "/content/drive/MyDrive/mri dataset.zip"
FILE_ID = "1pG_ITfqpBNMAQmg6Lvs6UjeQPZrW4VLL"

LOCAL_ZIP = "/content/mri_dataset.zip"
EXTRACT_DIR = "/content/mri_dataset"

if SOURCE == "drive":
    from google.colab import drive
    drive.mount("/content/drive")                      # approve the popup
    if ZIP_PATH is None:
        hits = glob.glob("/content/drive/MyDrive/**/*[mM][rR][iI]*.zip", recursive=True)
        print("zip files found:", hits)
        assert hits, ("No MRI zip found in My Drive. If someone shared it with you, right-click it in Drive "
                      "> Organize > Add shortcut to My Drive, or set SOURCE = 'link'.")
        ZIP_PATH = hits[0]
    shutil.copy(ZIP_PATH, LOCAL_ZIP)                   # local disk is much faster than Drive
else:
    import gdown
    gdown.download(id=FILE_ID, output=LOCAL_ZIP, quiet=False)

assert zipfile.is_zipfile(LOCAL_ZIP), "The file isn't a valid zip (incomplete copy, or the link isn't shared publicly)."

# ---- extract fresh every time ----
shutil.rmtree(EXTRACT_DIR, ignore_errors=True)
with zipfile.ZipFile(LOCAL_ZIP) as z:
    z.extractall(EXTRACT_DIR)

# ---- build the table of images ----
IMG_EXTS = {".jpg", ".jpeg", ".png", ".bmp", ".tif", ".tiff"}
all_files = sorted(
    p for p in Path(EXTRACT_DIR).rglob("*")
    if p.suffix.lower() in IMG_EXTS and "__MACOSX" not in p.parts
)

def folder_split(p):
    parts = [x.lower() for x in p.relative_to(EXTRACT_DIR).parts]
    if any(x in ("testing", "test") for x in parts):
        return "test"
    if any(x in ("training", "train") for x in parts):
        return "train"
    return "none"

df = pd.DataFrame({
    "path": all_files,
    "label": [p.parent.name.lower() for p in all_files],
    "folder_split": [folder_split(p) for p in all_files],
})

assert len(df) > 0, "No images found. Check the zip's contents."
assert df["label"].nunique() == 4, f"Expected 4 class folders, found: {sorted(df['label'].unique())}"

print(f"{len(df)} images found")
df.head()
```

```python
IMG_SIZE = 128          # increase to 224 for more detail (slower, more memory)

def crop_brain(img, thresh=10):
    # crop away the black border around the head
    mask = (img > thresh).astype(np.uint8)
    mask = cv2.morphologyEx(mask, cv2.MORPH_OPEN, np.ones((5, 5), np.uint8))
    cnts, _ = cv2.findContours(mask, cv2.RETR_EXTERNAL, cv2.CHAIN_APPROX_SIMPLE)
    if not cnts:
        return img
    x, y, w, h = cv2.boundingRect(max(cnts, key=cv2.contourArea))
    return img[y:y + h, x:x + w]

def resize_pad(img, size=IMG_SIZE):
    # resize keeping the aspect ratio, then pad with zeros to a square
    h, w = img.shape
    s = size / max(h, w)
    img = cv2.resize(img, (max(1, int(w * s)), max(1, int(h * s))), interpolation=cv2.INTER_AREA)
    ph, pw = size - img.shape[0], size - img.shape[1]
    return np.pad(img, ((ph // 2, ph - ph // 2), (pw // 2, pw - pw // 2)))

def preprocess(path, size=IMG_SIZE):
    img = cv2.imread(str(path), cv2.IMREAD_GRAYSCALE)          # one channel
    if img is None:
        raise ValueError(f"Could not read image: {path}")
    img = crop_brain(img)
    img = resize_pad(img, size).astype(np.float32) / 255.0      # scale to [0, 1]
    brain = img > (10 / 255)
    if brain.any():                                             # z-score using brain pixels only
        m, s = img[brain].mean(), img[brain].std()
        img = (img - m) / (s + 1e-8)
    return img.astype(np.float32)

# 1) remove exact duplicates (the first copy in sorted path order is kept)
df_clean = df.drop_duplicates("md5").reset_index(drop=True)
print(f"Removed {len(df) - len(df_clean)} duplicate files, {len(df_clean)} images remain")

# 2) encode labels as integers
class_names = sorted(df_clean["label"].unique())
label_to_idx = {c: i for i, c in enumerate(class_names)}
df_clean["y"] = df_clean["label"].map(label_to_idx)

# 3) split
if "test" in set(df_clean["folder_split"]) and "train" in set(df_clean["folder_split"]):
    train_val_df = df_clean[df_clean["folder_split"] != "test"]
    test_df = df_clean[df_clean["folder_split"] == "test"]
    train_df, val_df = train_test_split(
        train_val_df, test_size=0.15, stratify=train_val_df["y"], random_state=SEED)
    print("Using the dataset's own Testing folder as the test set")
else:
    train_df, tmp_df = train_test_split(
        df_clean, test_size=0.30, stratify=df_clean["y"], random_state=SEED)
    val_df, test_df = train_test_split(
        tmp_df, test_size=0.50, stratify=tmp_df["y"], random_state=SEED)
    print("No Training/Testing folders found: made a stratified 70/15/15 split")

# 4) build the arrays
def build_arrays(frame):
    X = np.stack([preprocess(p) for p in tqdm(frame["path"], leave=False)])[..., None]
    y = frame["y"].to_numpy()
    return X, y

X_train, y_train = build_arrays(train_df)
X_val, y_val = build_arrays(val_df)
X_test, y_test = build_arrays(test_df)
print("train:", X_train.shape, "| val:", X_val.shape, "| test:", X_test.shape)

# 5) class weights
weights = compute_class_weight("balanced", classes=np.arange(len(class_names)), y=y_train)
class_weight = dict(enumerate(weights))
print("class weights:", {class_names[i]: round(float(w), 2) for i, w in class_weight.items()})

# 6) augmentation (active only during training)
data_augmentation = keras.Sequential([
    layers.RandomFlip("horizontal"),
    layers.RandomRotation(0.04, fill_mode="nearest"),            # about +-14 degrees
    layers.RandomTranslation(0.05, 0.05, fill_mode="nearest"),
    layers.RandomZoom(0.1, fill_mode="nearest"),
], name="augmentation")

# visual check: original -> preprocessed -> augmented
idx = np.random.choice(len(train_df), 6, replace=False)
fig, axes = plt.subplots(3, 6, figsize=(15, 8))
for j, i in enumerate(idx):
    raw = cv2.imread(str(train_df["path"].iloc[i]), cv2.IMREAD_GRAYSCALE)
    pre = X_train[i]
    aug = data_augmentation(pre[None], training=True)[0].numpy()
    for r, (im, t) in enumerate([(raw, "original"), (pre[..., 0], "preprocessed"), (aug[..., 0], "augmented")]):
        axes[r, j].imshow(im, cmap="gray")
        axes[r, j].axis("off")
        axes[r, j].set_title(f"{class_names[y_train[i]]}\n{t}" if r == 0 else t, fontsize=9)
plt.tight_layout()
plt.show()
```

```python
def conv_block(x, filters):
    x = layers.Conv2D(filters, 3, padding="same", use_bias=False)(x)
    x = layers.BatchNormalization()(x)
    x = layers.ReLU()(x)
    x = layers.Conv2D(filters, 3, padding="same", use_bias=False)(x)
    x = layers.BatchNormalization()(x)
    x = layers.ReLU()(x)
    return layers.MaxPooling2D()(x)

inputs = keras.Input(shape=(IMG_SIZE, IMG_SIZE, 1))
x = data_augmentation(inputs)
for f in (32, 64, 128, 256):
    x = conv_block(x, f)
x = layers.GlobalAveragePooling2D()(x)
x = layers.Dropout(0.4)(x)
x = layers.Dense(128, activation="relu")(x)
x = layers.Dropout(0.3)(x)
outputs = layers.Dense(len(class_names), activation="softmax")(x)

model = keras.Model(inputs, outputs, name="brain_tumor_cnn")
model.summary()
```

```python
model.compile(
    optimizer=keras.optimizers.Adam(learning_rate=1e-3),
    loss="sparse_categorical_crossentropy",
    metrics=["accuracy"],
)

callbacks = [
    keras.callbacks.EarlyStopping(monitor="val_loss", patience=8, restore_best_weights=True),
    keras.callbacks.ReduceLROnPlateau(monitor="val_loss", factor=0.5, patience=3, min_lr=1e-5),
]
print("Model compiled.")
```

```python
EPOCHS = 40
BATCH_SIZE = 32

history = model.fit(
    X_train, y_train,
    validation_data=(X_val, y_val),
    epochs=EPOCHS,
    batch_size=BATCH_SIZE,
    class_weight=class_weight,
    callbacks=callbacks,
)
```

```python
test_loss, test_acc = model.evaluate(X_test, y_test, verbose=0)
print(f"Test loss: {test_loss:.4f} | Test accuracy: {test_acc:.4f}\n")

y_prob = model.predict(X_test, verbose=0)
y_pred = y_prob.argmax(axis=1)
label_ids = np.arange(len(class_names))

print(classification_report(y_test, y_pred, labels=label_ids, target_names=class_names, digits=3, zero_division=0))
report = classification_report(y_test, y_pred, labels=label_ids, target_names=class_names,
                               output_dict=True, zero_division=0)

cm = confusion_matrix(y_test, y_pred, labels=label_ids)
fig, ax = plt.subplots(figsize=(5.5, 5))
ConfusionMatrixDisplay(cm, display_labels=class_names).plot(ax=ax, cmap="Blues", colorbar=False)
plt.xticks(rotation=30)
plt.title("Confusion matrix (test set)")
plt.tight_layout()
plt.show()

f1s = {c: report[c]["f1-score"] for c in class_names}
best, worst = max(f1s, key=f1s.get), min(f1s, key=f1s.get)
print(f"Strongest class by F1: {best} ({f1s[best]:.3f}) | weakest: {worst} ({f1s[worst]:.3f})")

off_diag = cm.copy()
np.fill_diagonal(off_diag, 0)
i, j = np.unravel_index(off_diag.argmax(), off_diag.shape)
if off_diag[i, j] > 0:
    print(f"Most common mistake: true '{class_names[i]}' predicted as '{class_names[j]}' ({off_diag[i, j]} images)")
```

```python
hist = history.history
ep = range(1, len(hist["loss"]) + 1)

fig, axes = plt.subplots(1, 2, figsize=(11, 4))
axes[0].plot(ep, hist["accuracy"], label="train")
axes[0].plot(ep, hist["val_accuracy"], label="validation")
axes[0].set_title("Accuracy")
axes[0].set_xlabel("epoch")
axes[0].legend()

axes[1].plot(ep, hist["loss"], label="train")
axes[1].plot(ep, hist["val_loss"], label="validation")
axes[1].set_title("Loss")
axes[1].set_xlabel("epoch")
axes[1].legend()
plt.tight_layout()
plt.show()

best_epoch = int(np.argmin(hist["val_loss"])) + 1
gap = hist["accuracy"][-1] - hist["val_accuracy"][-1]
print(f"Epochs run: {len(ep)} | best epoch (lowest val loss): {best_epoch}")
print(f"Last epoch: train acc {hist['accuracy'][-1]:.3f}, val acc {hist['val_accuracy'][-1]:.3f}, gap {gap:+.3f}")
```
